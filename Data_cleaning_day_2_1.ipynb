```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv("DKHousingPricesSample100k.csv")
print(df.shape)
print(df.head())
print(df.describe())
```

```python
print(df.isnull().sum())
print(df.dtypes)
```

```python
# Fill missing values (NaNs) in the categorical 'city' column with the placeholder 'unknown'
df['city'] = df["city"].fillna("unknown")

# Calculate the mean value for the mortgage credit bond yield column
mean_value = df["yield_on_mortgage_credit_bonds%"].mean()

# Fill missing values in 'yield_on_mortgage_credit_bonds%' with the calculated mean
df["yield_on_mortgage_credit_bonds%"] = df["yield_on_mortgage_credit_bonds%"].fillna(mean_value)

# Calculate the mean value for the Danish annual inflation rate column
dk_mean = df["dk_ann_infl_rate%"].mean()

# Fill missing values in 'dk_ann_infl_rate%' with its calculated mean
df["dk_ann_infl_rate%"] = df["dk_ann_infl_rate%"].fillna(dk_mean)

# Verify that there are no remaining missing values in the dataset
print(df.isnull().sum())
```
